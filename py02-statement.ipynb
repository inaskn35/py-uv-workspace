```
  연산자 (operator) : 어떠한 기능을 수행하는 기호(+, -, *, /, %, >, <, >=, <=, ==, != ...)
  피연산자(operand) : 연산자의 작업 대상 ( 변수, 상수, 리터널, 수식...)
 
  연산자의 종류
 1. 산술연산자 : +, -, *, /,//, %
 2. 비교연산자 : >, <, >=, <=, ==, !=
 3. 논리연산자 : and, or, not
 4. 대입연산자 : =, +=, -=, *=, /=,...
 5. 식별연산자 : is, is not
```





````python
# 산술 연산자
#- 산술 연산자는 수학적 연산을 수행
#- +, -, *, /, %, **, //



print(10 / 4) # 실수 나눗셈을 수행
print(10 // 4) # 정수 나눗셈을 수행


print(2 ** 4) # 2의 4제곱 - 거듭제곱


# %는 정수나눗셈의 나머지결과를 얻는 연산자입니다.
print(4 % 3)
print(27 % 5)


print("-" * 40) #선그어줄때 사용


n1 = int(input("첫번째 정수: "))
n2 = int(input("두번째 정수: "))


print(n1, "과", n2, "의 나눗셈의 몫은", n1//n2,
      "이고, 나머지는", n1%n2, "입니다.")
````

````python
#  비교연산자
 # -  <       >       <=      >=      ==      !=
 # - 결과는 논리값 : True , False

print(3 < 2)
print(3 > 2)
print(3 == 2)
print(3 != 2)
````

````python
# 할당 연산자 (Assignment Operators)
#- 할당 연산자는 변수에 값을 할당하는 데 사용
#- =, +=, -=, *=, /= ...


a = 5
a += 3  # a는 이제 8 전자가 a = a+3 보다 빠르다. 
#전자는 어디에 할당될지를 먼저 알고 연산하는데, 후자는 연산을 하고 나서 어디에 할당할지를 알게되기에 속도차이
print(a)  # 8
````

````python
# 식별 연산자 (Identity Operators)
- 식별 연산자는 두 객체가 같은 객체인지를 확인한다.
- is : 두 객체가 같은 객체인지 확인
- is not 두 객체가 다른 객체인지 확인
- 결과 : True, False
````

````python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b)  # True (a와 b는 같은 객체)
print(a is c)  # False (a와 c는 같은 값을 가지지만 다른 객체)
````

````python
#   멤버 연산자 (Membership Operators)
#- 멤버 연산자는 시퀀스 내에 값이 존재하는지 확인한다.
#- in : 특정값이 멤버의 값인지 확인
#- in not : 특정값이 멤버의 값인지 아닌지 확인
#- 결과 : True, False


a = [1, 2, 3]
print(2 in a)  # True
print(4 not in a)  # True
````

````python
## 반복문 : while
반복문에서 else는 반복문이 완벽하게 끝났을 때 실행된다.  
대부분의 프로그래밍 언어는 while과 for을 지원한다.  
else
  - if ~ else    
  - while ~ else  
  - try ~ except ~ else ~ finally

```
while 조건식:
  수행할 문장
else:
  수행할 문장
```
````

````python
# while문이 완벽이  돌았으므로 end가 출력된다.
a = 10
while  a > 5 :
  a -= 1  # a = a - 1
  print(a)
  if a==5:
    break #반복문을 강제적으로 빠져나올때 사용한다. 즉, 반복문을 중단시킨다.
           #while 단계에서 아예 빠져나오는거라, else를 수행하지 않는다. 
else:
  print('end')
````

````python
#  continue는 다음 루프(조건식)로 바로 넘어간다.
# 반복문을 중단하지는 않으므로 else문이 실행된다.
a = 10 #5
while a > 5:
  a -= 1
  if a % 2 == 0 :
    continue   # 이동  True값이 나왔을때, 아래를 수행하지 않고 다시 while로 이동, False이면 print(a)
  print(a)
else:
  print('end')
````

````python
## enumerate 함수
 - enumerate()함수는 인덱스와 객체를 동시에 저장할 수 있는 특징을 가진 for문을 좀더 효율적으로 사용하기 위한 함수이다.
 - enumerate()함수는 반복가능객체(list, tuple, str)을 입력으로 받아 인덱스를 포함하는 enumerate 객체를 리턴한다.
 - enumerate([순서자료형], [시작값=0])의 형식을 가지며 반복가능객체은 반복자가 가능한 객체가 입력되고 시작값은 입력시 해당수 부터 증가하며, 생략했을 경우 기본값은 0 이다.
````

````python
for k in enumerate(['apple','python','microsoft']):
  print(k)
````

````python
for k , v in enumerate(['apple','python','microsoft']):
  print(k, v)
````

````python
#enumerate (oterable, start) => start는 생략이 가능하고, 생략한다면 default는 0이다. 인덱스가 0부터 시작한다는 의미
for k , v in enumerate(['apple','python','microsoft'], 1):
  print(k, v)
````

````python
a = {'one':10, 'two':20}
for i, k in enumerate(a):
  print(i, a[k])
````

````python
a = {'one':10, 'two':20}
for _, k in enumerate(a):
  print( a[k])
````

````python
### zip
#- 매개인자로 여러 목록을 받아 각 목록의 요소를 튜플에 담아 목록에 저장한다. 튜플로나온다.
# 짝 숫자가 안맞으면 짧은것에 맞춰지고, 나머지는 버려진다

a = [1, 2, 3, 4, 5]
b = ['a','b','c','d','f']


for x in zip(a,b):
  print(x)
````

````python
a = [1, 2, 3, 4, 5]
b = ['a','b','c','d','f']


for i, x in enumerate (zip(a,b)):
  print(i, x)
````

````python
a = [1, 2, 3, 4, 5]
b = ['a','b','c','d','f']


for i, (k, x) in enumerate (zip(a,b)):
  print(i, k, x)
````

````python
## Comprehension
 - comprehension이란 iterable한 오브젝트를 생성하기 위한 방법 중 하나이다.
   - List Comprehension(LC)
   - Set Comprehension(SC)
   - Dict Comprehension(DC)

List Comprehension
List 정의시 값을 정하지 않고 호출시 리스트 내의 값들이 처리되도록 구성
A = [표현식 for 항목 in 반복가능객체(iterable)]
A = [표현식 for 항목 in 반복가능객체(iterable) if 논리식]
A = [표현식 for 항목1 in 반복가능객체(iterable)1 for 항목2 in 반복가능객체(iterable)2 if 논리식]
````

````python
s = [ ]
 for x in range(10)
  s.append(x**2)
print(s)
````

````python
s = [ x**2 for x in range(10) ] #위의 방식을 간단히 한줄의 방식으로 표현해줄 수 있다. 
print(s)
````

````python
s = [ x**2 for x in range(10) if x % 2 ==0]
print(s)
````

````python
s = [ x if x%2 == 0   else -x for x in range(5)]
print(s)
````
