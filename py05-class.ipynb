````python
Book(title=None:str, author=None:str, page=None:int)  
+setter & getter  
+toString():str 
[출력결과]
파이썬의정석의 저자는 김파이썬이고 150페이지로 구성됨
책이름 :  파이썬 입문
책저자 :  박북스
페이지 :  200 페이지
파이썬 입문의 저자는 박북스이고 200페이지로 구성됨
````

````python
class Book:

    def __init__(self, title=None, author=None, page=None):
        self.__title = title
        self.__author = author
        self.__page = page

    def getTitle(self):
        return self.__title
    
    def setTitle(self, title):
        self.__title = title
    
    def getAuthor(self):
        return self.__author

    def setAuthor(self, author):
        self.__author = author

    def getPage(self):
        return self.__page

    def setPage(self, page):
        self.__page = page

    def toString(self):
        return f"{self.__title}의 저자는 {self.__author}이고 {self.__page}페이지로 구성됨"
    
class BookRun:  #실행할 클래스
    def process(self):
        bk1 = Book()
        bk1.setTitle('파이썬의 정석')
        bk1.setAuthor('김파이썬')
        bk1.setPage(150)
        print(bk1.toString())
        #print(bk1._Book__title)
        print(dir(bk1))

        bk2 = Book('파이썬 입문', '박북스', 200)
        print('책이름 : {}'.format(bk2.getTitle()))
        print('책저자 : {}'.format(bk2.getAuthor()))
        print('페이지 : {}'.format(bk2.getPage()))
        print(bk2.toString())


if __name__ == '__main__':
    bn = BookRun()
    bn.process()
````

````python
from abc import ABC, abstractmethod


class Shape(ABC):


    def __init__(self, width, height):
        self.width = width
        self.height = height


    @abstractmethod
    def area(self):
      pass    




class Rect(Shape):
   
    def __init__(self, width, height ):
        super().__init__(width, height)


    def area(self):
      return self.width * self.height  


   
class Triangle(Shape):
   
    def __init__(self, width, height ):
        super().__init__(width, height)


    def area(self):
      return self.width * self.height * 0.5       
````
